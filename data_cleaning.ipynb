```python
print(df["age"].unique())
print(f'min: {df["age"].min()}')
print(f'max: {df["age"].max()}')
```

```python
df = df[df["age"] < 100]
```

```python
plt.figure(figsize=(5,3))
v = df["age"].value_counts()
plt.bar(v.index, v)
plt.title("Age distribution")
plt.show()
```
